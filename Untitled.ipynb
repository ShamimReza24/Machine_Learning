```python
# ==========================================
# 1. Import libraries
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score


# ==========================================
# 2. Load dataset
# ==========================================

train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")


# ==========================================
# 3. Check the dataset
# ==========================================

print("Train shape:", train.shape)
print("Test shape:", test.shape)

print("\nTrain columns:")
print(train.columns)

print("\nFirst 5 rows:")
print(train.head())

print("\nData types:")
print(train.dtypes)


# ==========================================
# 4. Check missing/null values
# ==========================================

print("\nNull values:")
print(train.isnull().sum())

print("\nNull values in selected features:")
print(train[[
    "OverallQual",
    "GrLivArea",
    "GarageCars",
    "TotalBsmtSF",
    "1stFlrSF",
    "FullBath",
    "YearBuilt"
]].isnull().sum())


# ==========================================
# 5. Select 7 important features
# ==========================================

features = [
    "OverallQual",
    "GrLivArea",
    "GarageCars",
    "TotalBsmtSF",
    "1stFlrSF",
    "FullBath",
    "YearBuilt"
]


# ==========================================
# 6. Handle missing values
# ==========================================

# Fill missing values with 0
train[features] = train[features].fillna(0)
test[features] = test[features].fillna(0)


# ==========================================
# 7. Create X and Y
# ==========================================

X = train[features]
Y = train["SalePrice"]


# ==========================================
# 8. Split training data
# ==========================================

X_train, X_val, Y_train, Y_val = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)


# ==========================================
# 9. Create Linear Regression model
# ==========================================

model = LinearRegression()


# ==========================================
# 10. Train the model
# ==========================================

model.fit(X_train, Y_train)


# ==========================================
# 11. Predict validation data
# ==========================================

Y_pred = model.predict(X_val)


# ==========================================
# 12. Check R²
# ==========================================

r2 = r2_score(Y_val, Y_pred)

print("\nR² Score:", r2)


# ==========================================
# 13. Check Mean Squared Error
# ==========================================

mse = mean_squared_error(Y_val, Y_pred)

print("Mean Squared Error:", mse)


# ==========================================
# 14. Predict one house manually
# ==========================================

house = [[
    8,       # OverallQual
    2000,    # GrLivArea
    2,       # GarageCars
    1000,    # TotalBsmtSF
    1000,    # 1stFlrSF
    2,       # FullBath
    2010     # YearBuilt
]]

prediction = model.predict(house)

print("\nPredicted price:", prediction[0])


# ==========================================
# 15. Train on ALL training data
# ==========================================

model.fit(X, Y)


# ==========================================
# 16. Prepare test data
# ==========================================

X_test = test[features]


# ==========================================
# 17. Predict test prices
# ==========================================

test_prediction = model.predict(X_test)


# ==========================================
# 18. Create submission file
# ==========================================

submission = pd.DataFrame({
    "Id": test["Id"],
    "SalePrice": test_prediction
})


# ==========================================
# 19. Save submission.csv
# ==========================================

submission.to_csv("submission.csv", index=False)


# ==========================================
# 20. Check submission
# ==========================================

print("\nSubmission:")
print(submission.head())

print("\nSubmission shape:")
print(submission.shape)

print("\nSubmission null values:")
print(submission.isnull().sum())

print("\nsubmission.csv created successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Read CSV
df = pd.read_csv('submission (1).csv')

# 2. Parse PredictionString into flat records
records = []
for _, row in df.iterrows():
    pstr = str(row['PredictionString']).strip()
    if pstr:
        parts = pstr.split()
        for i in range(0, len(parts), 6):
            if i + 5 < len(parts):
                records.append({
                    'image_id': row['image_id'],
                    'class_id': int(parts[i]),
                    'confidence': float(parts[i+1]),
                    'x': float(parts[i+2]),
                    'y': float(parts[i+3]),
                    'w': float(parts[i+4]),
                    'h': float(parts[i+5]),
                    'cx': float(parts[i+2]) + float(parts[i+4])/2,
                    'cy': float(parts[i+3]) + float(parts[i+5])/2
                })

boxes_df = pd.DataFrame(records)

# 3. Create Dashboard Figure
fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# Panel A: Objects per Class
sns.countplot(data=boxes_df, x='class_id', ax=axes[0,0], palette='viridis')
axes[0,0].set_title('Total Objects per Class ID', fontsize=12, fontweight='bold')
axes[0,0].set_xlabel('Class ID')

# Panel B: Spatial Center Heatmap
axes[0,1].hexbin(boxes_df['cx'], boxes_df['cy'], gridsize=30, cmap='YlOrRd', mincnt=1)
axes[0,1].set_title('Spatial Distribution of Object Centers (cx, cy)', fontsize=12, fontweight='bold')
axes[0,1].invert_yaxis()  # Image origin is top-left

# Panel C: Bounding Box Width vs Height
sns.scatterplot(data=boxes_df, x='w', y='h', hue='class_id', palette='tab10', alpha=0.6, ax=axes[1,0], s=25)
axes[1,0].set_title('Bounding Box Aspect Ratio (Width vs Height)', fontsize=12, fontweight='bold')

# Panel D: Confidence Distribution by Class
sns.boxplot(data=boxes_df, x='class_id', y='confidence', ax=axes[1,1], palette='coolwarm')
axes[1,1].set_title('Confidence Score Distribution by Class', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# 1. Load the CSV file
df = pd.read_csv('submission (1).csv')

# 2. Select the specific image row (first row in this case)
row_index = 0
target_row = df.iloc[row_index]
image_id = target_row['image_id']

# 3. Parse prediction string
prediction_str = str(target_row['PredictionString']).strip()
boxes = []

if prediction_str:
    parts = prediction_str.split()
    for i in range(0, len(parts), 6):
        boxes.append({
            'class_id': int(parts[i]),
            'confidence': float(parts[i+1]),
            'x': float(parts[i+2]),
            'y': float(parts[i+3]),
            'w': float(parts[i+4]),
            'h': float(parts[i+5])
        })

# 4. Initialize Plot Canvas
fig, ax = plt.subplots(figsize=(10, 10))
ax.set_xlim(0, 1)
ax.set_ylim(1, 0)  # Invert Y-axis for standard image coordinates
ax.set_facecolor('#1e1e1e')  # Dark canvas background

# Color palette mapped across class IDs
colors = plt.cm.get_cmap('tab20', 13)

# 5. Draw Bounding Boxes and Labels
for b in boxes:
    cls_color = colors(b['class_id'])
    
    # Draw Rectangle
    rect = patches.Rectangle(
        (b['x'], b['y']), b['w'], b['h'],
        linewidth=2,
        edgecolor=cls_color,
        facecolor='none'
    )
    ax.add_patch(rect)
    
    # Draw Class Label with Background Box
    label_text = f"Cls {b['class_id']} ({b['confidence']:.2f})"
    ax.text(
        b['x'], b['y'] - 0.01, label_text,
        color='white',
        fontsize=9,
        bbox=dict(
            facecolor=cls_color,
            alpha=0.8,
            pad=1.5,
            edgecolor='none'
        )
    )

# 6. Styling and Labels
ax.set_title(f"Predicted Bounding Boxes for:\n{image_id}", fontsize=12, fontweight='bold', pad=15)
ax.set_xlabel("Normalized X", fontsize=11)
ax.set_ylabel("Normalized Y", fontsize=11)
ax.tick_params(axis='both', labelsize=11)
plt.grid(True, linestyle=':', alpha=0.3)

plt.tight_layout()
plt.show()
```
